```python
import numpy as np
import pandas as pd

x1 = np.array([1,2,3,7,8,9])
x2 = np.array([2,3,4,6,7,8])
x3 = np.array([5,4,3,3,2,1])
y = np.array([0,0,0,1,1,1])
```

```python
print(np.corrcoef(x1,x2)[0,1]) #pandas .corr()
```

```python
rx1x2 = abs(np.corrcoef(x1,x2)[0,1])
rx1x3 = abs(np.corrcoef(x1,x3)[0,1])
rx2x3 = abs(np.corrcoef(x2,x3)[0,1])
rx1y = abs(np.corrcoef(x1,y)[0,1])
rx2y = abs(np.corrcoef(x2,y)[0,1])
rx3y = abs(np.corrcoef(x3,y)[0,1])
```

```python
rcf12 = sum([rx1y,rx2y])/2
rcf13 = sum([rx1y,rx3y])/2
rcf23 = sum([rx2y,rx3y])/2
```

```python
def merit(k,rcf,rff):
  return (k*rcf)/(k+((k*(k-1))*abs(rff)))**0.5
```

```python
m12 = merit(2,rcf12,rx1x2)
m23 = merit(2,rcf23,rx2x3)
m13 = merit(2,rcf13,rx1x3)
```

```python
if (m12 > m23 and m12 >m13):
  print(f"Selected features x1 and x2 m12 = {m12}")
elif (m23>m13):
  print(f"Selected features x2 and x3 m23 = {m23}")
else:
  print(f"Selected features x1 and x3 m13 = {m13}")
```
