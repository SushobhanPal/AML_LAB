```python
n = int(input(("Enter number of features in X: ")))
x =[]
y =[]

for i in range(n):
    x.append(float(input("Enter the value of X: ")))
    y.append(float(input("Enter the value of Y: ")))
```

```python
sum_x = sum(x)
sum_y = sum(y)
```

```python
xy=[]
x2=[]
```

```python
for i in range(n):
  xy.append(x[i]*y[i])
  x2.append(x[i]*x[i])
```

```python
sum_xy = sum(xy)
sum_x2 = sum(x2)
```

```python
slope = ((n*sum_xy) - (sum_x*sum_y)) / ((n*sum_x2) - (sum_x*sum_x))
```

```python
print(slope)
```

```python
intercept = (sum_y - slope * sum_x) / n
print(intercept)
```

```python
print(f"Y = {slope} * X + {intercept}")
```

```python
y_pred = []
for i in range(n):
  y_pred.append(slope*x[i] + intercept)
```

```python
y_pred
```

```python
error = []
for i in range(n):
  error.append(abs(y[i]-y_pred[i]))
```

```python
error
```

```python
sum_error = sum(error)
```

```python
MAE = sum_error/n
print(MAE)
```

```python
error_square = []
for i in range(n):
  error_square.append(error[i]**2)
```
