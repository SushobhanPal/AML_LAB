```python
import pandas as pd


data = {
    'Rainfall': ['High', 'High', 'High', 'Low', 'Low', 'Low'],
    'Soiltype': ['Loamy', 'Sandy', 'Clay', 'Loamy', 'Sandy', 'Clay'],
    'Yield': ['Good', 'Good', 'Good', 'Poor', 'Poor', 'Good']
}
df = pd.DataFrame(data)



def fit_one_r(df, target_col):
    features = [col for col in df.columns if col != target_col]
    best_feature = None
    best_rules = {}
    min_errors = float('inf')

    for feature in features:
        feature_rules = {}
        feature_errors = 0


        for value in df[feature].unique():

            subset = df[df[feature] == value]


            majority_class = subset[target_col].value_counts().idxmax()
            feature_rules[value] = majority_class


            errors = len(subset[subset[target_col] != majority_class])
            feature_errors += errors

        print(f"Feature '{feature}' has total errors: {feature_errors}")


        if feature_errors < min_errors:
            min_errors = feature_errors
            best_feature = feature
            best_rules = feature_rules

    return best_feature, best_rules, min_errors

best_feat, rules, total_errors = fit_one_r(df, 'Yield')


print(f"Best Predictor: {best_feat}")
print("Generated Rules:")
for val, prediction in rules.items():
    print(f"{best_feat} == '{val}'=> Yield = '{prediction}'")
```
